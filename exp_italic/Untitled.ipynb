```python
from PIL import Image, ImageDraw, ImageFont
import matplotlib.pyplot as plt
import os

PATH_FONTS = "C:/Users/Маша/PycharmProjects/FontEmb/variable_fonts"
# output_dir = "font_comparison"

text = "фывапролджукенгшщ"  
width = 800
height = 200 
bg_color = (255, 255, 255)  
text_color = (0, 0, 0)  
font_size = 40

all_fonts = os.listdir(PATH_FONTS)
nonitalic_fonts = [f for f in all_fonts if '_italic' not in f]
italic_fonts = [f for f in all_fonts if '_italic' in f]

for font in nonitalic_fonts:

    img = Image.new('RGB', (width, height), bg_color)
    draw = ImageDraw.Draw(img)

    italic_font = font.replace('.ttf', '_italic.ttf')
    
    regular = ImageFont.truetype(os.path.join(PATH_FONTS, font), font_size)
    italic = ImageFont.truetype(os.path.join(PATH_FONTS, italic_font), font_size)
                      
    draw.text((50, 80), text, fill=text_color, font=regular)
                        
    draw.text((450, 80), text, fill=text_color, font=italic)
                      
    draw.line([(400, 0), (400, 200)], fill=(200, 200, 200), width=2)

    # img.save(f"{output_dir}/{nonitalic_font}_compare.png")

    plt.figure(figsize=(10, 5 * len(nonitalic_fonts)))
    plt.imshow(img)
    plt.axis('off')
    plt.title(font.replace('.ttf', ''))
    plt.show()
```
