```python
simclr_transform = T.Compose([
    T.RandomResizedCrop(32),
    T.RandomHorizontalFlip(),
    T.RandomApply([T.ColorJitter(0.4,0.4,0.4,0.1)], p=0.8),
    T.RandomGrayscale(p=0.2),
    T.GaussianBlur(kernel_size=3),
    T.ToTensor(),
    T.Normalize((0.4914, 0.4822, 0.4465), (0.247, 0.243, 0.261))
])
```

```python
import os
from PIL import Image
import random 
class SimCLRDataset(Dataset):
    def __init__(self, transform):
        self.image_dir = os.path.join(os.getcwd(), 'contr_dataset')
        self.images = [f for f in os.listdir(self.image_dir) 
                      if os.path.isfile(os.path.join(self.image_dir, f))]
        self.transform = transform

    def __getitem__(self, index):
        image_path = os.path.join(self.image_dir, self.images[index])
        image = Image.open(image_path).convert('RGB')
        crop1 = random.randint(0, 16000-45)
        crop2 = random.randint(0, 16000-45)
        image1 = image.crop((crop1, 0, crop1+40, 40))
        image2 = image.crop((crop2, 0, crop2+40, 40))
        xi = self.transform(image1)
        xj = self.transform(image2)
        return xi, xj

    def __len__(self):
        return len(self.images)
    
dataset = SimCLRDataset(simclr_transform)
sample = dataset[0] 
print(sample[0].shape, sample[1].shape)


```

```python
class SimCLRModel(nn.Module):
    def __init__(self, projection_dim=128):
        super().__init__()
        base_model = models.resnet18(weights=None) # ResNet без последнего FC слоя
        num_ftrs = base_model.fc.in_features # Получаем размер выходного вектора энкодера
        base_model.fc = nn.Identity()  
        self.encoder = base_model
        self.projection_head = nn.Sequential(
            nn.Linear(num_ftrs, 512),
            nn.ReLU(),
            nn.Linear(512, projection_dim)
        )

    def forward(self, x):
        h = self.encoder(x) # Получаем представление от энкодера
        z = self.projection_head(h) # Проецируем в пространство меньшей размерности
        return z
```

```python
def nt_xent_loss(z_i, z_j, temperature=0.5):
    z = torch.cat([z_i, z_j], dim=0) # Объединяем все представления в одну матрицу (удваиваем размер батча)
    z = F.normalize(z, dim=1) # L2-нормализация по вектору признаков

    similarity = torch.matmul(z, z.T) # [2*N, 2*N] матрица косинусных сходств
    N = z_i.shape[0] # исходный размер батча

    mask = (~torch.eye(2*N, dtype=bool)).to(z.device) # False на диагонали 
    sim = similarity / temperature
    exp_sim = torch.exp(sim) * mask # Обнуляем диагональ, чтобы не учитывать сходство вектора с самим собой

    positive_sim = torch.exp(F.cosine_similarity(z_i, z_j) / temperature)
    positives = torch.cat([positive_sim, positive_sim], dim=0) # позитивные пары (дублируем, так как каждая аугментация должна притягиваться к своей паре)

    denominator = exp_sim.sum(dim=1) # сумма по всем примерам
    loss = -torch.log(positives / denominator)
    return loss.mean() # Loss как Softmax: максимизировать схожесть положительных пар относительно всех остальных
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

contrastive_dataset = SimCLRDataset(simclr_transform)
train_loader = DataLoader(contrastive_dataset, batch_size=128, shuffle=True)

model = SimCLRModel().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=3e-4)

for epoch in range(30):
    model.train()
    total_loss = 0
    for x_i, x_j in tqdm(train_loader):
        x_i, x_j = x_i.to(device), x_j.to(device)
        z_i = model(x_i)
        z_j = model(x_j)

        loss = nt_xent_loss(z_i, z_j)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    print(f"Epoch {epoch+1} | Loss: {total_loss / len(train_loader):.4f}")
```

```python
for param in model.encoder.parameters():
    param.requires_grad = False

classifier = nn.Linear(512, 10).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(classifier.parameters(), lr=1e-3)

def get_features_and_labels(loader):
    features, labels = [], []
    with torch.no_grad():
        for x, y in loader:
            x = x.to(device)
            h = model.encoder(x)
            features.append(h.cpu())
            labels.append(y)
    return torch.cat(features), torch.cat(labels)
```

```python

train_dataset_eval = CIFAR10(root='./data', train=True, transform=simclr_transform)
train_loader_eval = DataLoader(train_dataset_eval, batch_size=256, shuffle=False)

test_dataset = CIFAR10(root='./data', train=False, transform=simclr_transform)
test_loader = DataLoader(test_dataset, batch_size=256, shuffle=False)

X_train, y_train = get_features_and_labels(train_loader_eval)
X_test, y_test = get_features_and_labels(test_loader)

X_train, y_train = X_train.to(device), y_train.to(device)
X_test = X_test.to(device)

classifier.train()
for epoch in range(20):
    for i in range(0, len(X_train), 256):
        x_batch = X_train[i:i+256]
        y_batch = y_train[i:i+256]
        optimizer.zero_grad()
        logits = classifier(x_batch)
        loss = criterion(logits, y_batch)
        loss.backward()
        optimizer.step()
    if (epoch + 1) % 5 == 0:
        print(f"Epoch {epoch+1} | Loss: {loss.item():.4f}")
```

```python
classifier.eval()
with torch.no_grad():
    logits = classifier(X_test)
    _, predicted = logits.max(1)
    acc = (predicted.cpu() == y_test).float().mean().item()
    print(f"Точность: {acc * 100:.2f}%")
```
